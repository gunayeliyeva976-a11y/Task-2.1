```python
from Bio import Entrez

Entrez.email= "gunayeliyeva976@gmail.com"
```

```python
handle=Entrez.esearch(
    db="nucleotide",
    term="GADPH[Gene] AND Homo Sapiens[Organisms]",
     retmax=1
)
record=Entrez.read(handle)

print(record["IdList"])
```

```python
id = record["IdList"][0]

fetch_handle = Entrez.efetch(
    db="nucleotide",
    id=id,
    rettype="Fasta",
    retmode="text"
)
sequence_data = fetch_handle.read()
print(sequence_data[:1000])
```

```python
from io import StringIO
from Bio import SeqIO

seq_record = SeqIO.read(
    StringIO(sequence_data),
    "fasta"
)

print("ID:",seq_record.id)
print("Length:",len(seq_record.seq))
print("First 102 bases:")
print(seq_record.seq[:102])
      
```

```python
dna=seq_record.seq[:102]
print(dna)
```

```python
g=dna.count("G")
c=dna.count("C")
a=dna.count("A")
t=dna.count("T")
print("A:",a)
print("G:",g)
print("C:",c)
print("T:",t)
```
